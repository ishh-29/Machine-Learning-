```python
def sigmoid(z):
    g=1/(1+np.exp(-z))
    return g
```

```python
temp_z=np.arange(-10,11)
y=sigmoid(temp_z)
np.set_printoptions(precision=3)
print("Input (z), Output (sigmoid(z))")
print(np.c_[temp_z,y])
```

```python
#Plotting 
fig,ax=plt.subplots(1,1,figsize=(5,3))
ax.plot(temp_z,y,c='b')
ax.set_title("Sigmoid Function")
ax.set_ylabel('Sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```
