Implementation And Utilization Of Cost Function For Logistic Regression

```python
#Importing Modules And Libraries
import numpy as np,matplotlib.pyplot as plt
%matplotlib widget
from lab_utils_common_class import plot_data,sigmoid,dlc
plt.style.use('./deeplearning.mplstyle')

```

```python
#Loading Data Set
X_train=np.array([[0.5,1.5],[1,1],[1.5,0.5],
                  [3,0.5],[2,2],[1,2.5]])
y_train=np.array([0,0,0,1,1,1])
#Plotting This Data
fig,ax=plt.subplots(1,1,figsize=(4,4))
plot_data(X_train,y_train,ax)
#Setting Both Axes To Be In Range [0,4]
ax.axis([0,4,0,3.5])
ax.set_ylabel('$x_1$',fontsize=12)
ax.set_xlabel('$x_0$',fontsize=12)
plt.show()
```

```python
#Computing The Cost Function For Logistic Regression
def logistic_cost(x,y,w,b):
    m=x.shape[0]
    cost=0.0
    for i in range(m):
        z=np.dot(w,x[i])+b
        f=sigmoid(z)
        cost+=-y[i]*np.log(f)-(1-y[i])*np.log(1-f)
    cost/=m
    return cost
```

```python
#Running The Function 
temp_w=np.array([1,1])
temp_b=-3
print(logistic_cost(X_train,y_train,temp_w,temp_b))
```

```python
#Plotting Decision Boundary For Different b Values
x0=np.arange(0,6)
#Plotting Two Decision Boundaries
x1=3-x0
x1_prime=4-x0

fig,ax=plt.subplots(1,1,figsize=(4,4))
ax.plot(x0,x1,c=dlc['dlblue'],label='$b=3$')
ax.plot(x0,x1_prime,c=dlc['dlmagenta'],label='$b=4$')
ax.axis([0,4,0,4])

#Plotting Original Data
plot_data(X_train,y_train,ax)
ax.axis([0,4,0,4])
ax.set_ylabel('$x_1$',fontsize=12)
ax.set_xlabel('$x_0$',fontsize=12)
plt.legend(loc="upper right")
plt.title("Decision Boundary")
plt.show()
#b=-4 Is A Worse Model For The Training Data
```

```python
#Using Cost Function To Reflect This
w_arr1=np.array([1,1])
b1=-3
w_arr2=np.array([1,1])
b2=-4

print("Cost For b=-3:",logistic_cost(X_train,y_train,w_arr1,b1))
print("Cost For b=-4:",logistic_cost(X_train,y_train,w_arr2,b2))
```
